```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import warnings

warnings.filterwarnings("ignore", category=FutureWarning)
sns.set_theme(style="whitegrid")

# Find the extracted CSV files
data_folder = Path.home() / "Downloads" / "archive"
csv_files = list(data_folder.glob("*.csv"))

if not csv_files:
    raise FileNotFoundError(
        f"No CSV files found in {data_folder}. "
        "Check your extracted dataset folder."
    )

datasets = {}

for file in csv_files:
    df_temp = pd.read_csv(file)
    df_temp.columns = df_temp.columns.str.strip()
    df_temp = df_temp.drop_duplicates().copy()

    # Clean column names and text values
    for col in df_temp.select_dtypes(include="object").columns:
        df_temp[col] = df_temp[col].str.strip()

    # Convert date column when present
    for col in df_temp.columns:
        if col.lower() == "date":
            df_temp[col] = pd.to_datetime(
                df_temp[col], errors="coerce", dayfirst=True
            )

    # Convert numeric columns safely
    for col in df_temp.columns:
        if any(word in col.lower() for word in
               ["unemployment rate", "employed",
                "participation rate", "longitude", "latitude"]):
            df_temp[col] = pd.to_numeric(
                df_temp[col], errors="coerce"
            )

    datasets[file.name] = df_temp

    print("\nDataset:", file.name)
    print("Rows and columns:", df_temp.shape)
    print("Duplicate rows remaining:", df_temp.duplicated().sum())
    print("Missing values:")
    print(df_temp.isnull().sum())

# Select the largest dataset for the main analysis
dataset_name = max(datasets, key=lambda name: len(datasets[name]))
df = datasets[dataset_name].copy()

# Standardize common column names
rename_map = {}
for col in df.columns:
    lower = col.lower()
    if "unemployment rate" in lower:
        rename_map[col] = "Unemployment_Rate"
    elif "labour participation rate" in lower or "labor participation rate" in lower:
        rename_map[col] = "Labour_Participation_Rate"
    elif "estimated employed" in lower:
        rename_map[col] = "Estimated_Employed"

df = df.rename(columns=rename_map)

# Remove rows missing the main unemployment measure
if "Unemployment_Rate" not in df.columns:
    raise ValueError("The dataset has no unemployment rate column.")

df = df.dropna(subset=["Unemployment_Rate"]).copy()

print("\nSelected dataset:", dataset_name)
print("Cleaned dataset shape:", df.shape)
print("\nCleaned dataset preview:")
display(df.head())
print("\nStatistical summary:")
display(df.describe(include="all").T)
```

```python
# FINAL ANALYSIS: UNEMPLOYMENT ANALYSIS IN INDIA

import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create an output folder
output_folder = "Unemployment_Project_Output"
os.makedirs(output_folder, exist_ok=True)

# Use the cleaned dataset already available in your notebook
final_df = df.copy()

# Standardize column names
final_df.columns = (
    final_df.columns.astype(str)
    .str.strip()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
    .str.replace("%", "", regex=False)
)

# Identify columns
print("Dataset shape:", final_df.shape)
print("\nColumns:", final_df.columns.tolist())
print("\nMissing values:\n", final_df.isnull().sum())

# Convert date column
if "Date" in final_df.columns:
    final_df["Date"] = pd.to_datetime(
        final_df["Date"], errors="coerce"
    )

# Convert numerical columns
numeric_cols = [
    "Unemployment_Rate",
    "Estimated_Unemployed",
    "Estimated_Employed",
    "Labour_Participation_Rate"
]

for col in numeric_cols:
    if col in final_df.columns:
        final_df[col] = pd.to_numeric(
            final_df[col], errors="coerce"
        )

# Remove duplicate rows and handle missing values
final_df = final_df.drop_duplicates()

for col in numeric_cols:
    if col in final_df.columns:
        final_df[col] = final_df[col].fillna(
            final_df[col].median()
        )

print("\nCleaned dataset shape:", final_df.shape)
print("\nStatistical Summary:")
display(final_df.describe(include="all"))

sns.set_theme(style="whitegrid")

# 1. Unemployment rate distribution
if "Unemployment_Rate" in final_df.columns:
    plt.figure(figsize=(9, 5))
    sns.histplot(
        final_df["Unemployment_Rate"].dropna(),
        bins=20, kde=True
    )
    plt.title("Distribution of Unemployment Rates in India")
    plt.xlabel("Unemployment Rate (%)")
    plt.ylabel("Number of Records")
    plt.tight_layout()
    plt.savefig(
        f"{output_folder}/01_unemployment_distribution.png",
        dpi=300
    )
    plt.show()

# 2. Average unemployment rate by region
if {"Region", "Unemployment_Rate"}.issubset(final_df.columns):
    region_avg = (
        final_df.groupby("Region")["Unemployment_Rate"]
        .mean().sort_values(ascending=False)
    )

    plt.figure(figsize=(12, 7))
    region_avg.plot(kind="bar")
    plt.title("Average Unemployment Rate by Region")
    plt.xlabel("Region")
    plt.ylabel("Average Unemployment Rate (%)")
    plt.xticks(rotation=75, ha="right")
    plt.tight_layout()
    plt.savefig(
        f"{output_folder}/02_region_comparison.png",
        dpi=300
    )
    plt.show()

    print("\nFive regions with the highest average rates:")
    display(region_avg.head(5).round(2))

    print("\nFive regions with the lowest average rates:")
    display(region_avg.tail(5).round(2))

# 3. Unemployment trend over time
if {"Date", "Unemployment_Rate"}.issubset(final_df.columns):
    time_data = final_df.dropna(
        subset=["Date", "Unemployment_Rate"]
    ).copy()

    time_data["Month"] = (
        time_data["Date"].dt.to_period("M").dt.to_timestamp()
    )

    monthly_avg = (
        time_data.groupby("Month")["Unemployment_Rate"]
        .mean().sort_index()
    )

    plt.figure(figsize=(12, 5))
    plt.plot(
        monthly_avg.index, monthly_avg.values,
        marker="o"
    )
    plt.title("Unemployment Trend Over Time in India")
    plt.xlabel("Month")
    plt.ylabel("Average Unemployment Rate (%)")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig(
        f"{output_folder}/03_monthly_trend.png",
        dpi=300
    )
    plt.show()

    print("\nMonthly unemployment summary:")
    display(monthly_avg.round(2).to_frame(
        "Average Unemployment Rate (%)"
    ))

# 4. Urban versus rural comparison
if {"Area", "Unemployment_Rate"}.issubset(final_df.columns):
    area_avg = final_df.groupby("Area")[
        "Unemployment_Rate"
    ].mean().sort_values(ascending=False)

    plt.figure(figsize=(7, 5))
    area_avg.plot(kind="bar")
    plt.title("Average Unemployment: Urban vs Rural")
    plt.xlabel("Area")
    plt.ylabel("Average Unemployment Rate (%)")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.savefig(
        f"{output_folder}/04_area_comparison.png",
        dpi=300
    )
    plt.show()

    print("\nArea-wise average unemployment:")
    display(area_avg.round(2))

# 5. Correlation between numerical variables
available_numeric = final_df.select_dtypes(
    include="number"
)

if available_numeric.shape[1] >= 2:
    plt.figure(figsize=(9, 6))
    sns.heatmap(
        available_numeric.corr(),
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )
    plt.title("Correlation Between Numerical Variables")
    plt.tight_layout()
    plt.savefig(
        f"{output_folder}/05_correlation_heatmap.png",
        dpi=300
    )
    plt.show()

# 6. Save cleaned data
final_df.to_csv(
    f"{output_folder}/cleaned_unemployment_data.csv",
    index=False
)

# 7. Generate project findings
findings = [
    "PROJECT: UNEMPLOYMENT ANALYSIS IN INDIA",
    "=" * 45,
    f"Total records after cleaning: {len(final_df)}",
    f"Total columns: {len(final_df.columns)}",
]

if "Region" in final_df.columns:
    findings.append(
        f"Number of regions: {final_df['Region'].nunique()}"
    )

if "Unemployment_Rate" in final_df.columns:
    findings.extend([
        f"Average unemployment rate: "
        f"{final_df['Unemployment_Rate'].mean():.2f}%",
        f"Minimum recorded rate: "
        f"{final_df['Unemployment_Rate'].min():.2f}%",
        f"Maximum recorded rate: "
        f"{final_df['Unemployment_Rate'].max():.2f}%"
    ])

if "Region" in final_df.columns and "Unemployment_Rate" in final_df.columns:
    findings.append(
        "Highest average regional unemployment rate: "
        + str(region_avg.index[0])
    )
    findings.append(
        "Lowest average regional unemployment rate: "
        + str(region_avg.index[-1])
    )

findings.extend([
    "",
    "CONCLUSION",
    "This project analyzes unemployment data in India using Python.",
    "Data cleaning and exploratory analysis were performed using Pandas.",
    "Visualizations were created using Matplotlib and Seaborn.",
    "The analysis compares regional unemployment, trends over time,",
    "and differences between urban and rural areas.",
    "The findings can help identify patterns for further research.",
    "",
    "Note: Results describe the supplied dataset and do not establish",
    "the causes of unemployment."
])

report_text = "\n".join(findings)

with open(
    f"{output_folder}/project_summary.txt",
    "w",
    encoding="utf-8"
) as file:
    file.write(report_text)

print("\n" + report_text)
print("\nPROJECT ANALYSIS COMPLETED SUCCESSFULLY!")
print("Output folder:", os.path.abspath(output_folder))
```
